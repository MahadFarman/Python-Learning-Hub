### INHERITANCE


```python
# SINGLE INHERITANCE

class company:
    company = "Microsoft"
    deparment = "IT"

    def status(self):
        print(f"The company is {self.company} and the department is {self.deparment}")

class employee(company):
    Name = "Mahad"
    salary = 200000

    def employee_status(self):
        print(f"The name of employee is {self.Name} and the salary of tis employee is {self.salary}")

a=employee()
a.status()
a.employee_status()
```

```python
# MULTIPLE INHERITANCE

class company:
    company = "Microsoft"
    deparment = "IT"

    def status(self):
        print(f"The company is {self.company} and the department is {self.deparment}")

class employee(company):
    Name = "Mahad"
    salary = 200000

    def employee_status(self):
        print(f"The name of employee is {self.Name} and the salary of tis employee is {self.salary}")

class experiance(employee, company):
    experiance_employee = "5 years "
    skills = "programming"

    def character(self):
        print(f"The experiance of this employee is {self.experiance_employee} and skills are {self.skills}")

a=experiance()
a.status()
a.employee_status()
a.character()
```

```python
# MULTI_LEVEL_INHERITANCE

class person:
    def per_info(self, name ="mahad", age = "16"):
        print(f"The person name is {name} and his age is {age}")

class employee(person):
    def emp_info(self, emp_id = 1234):
        print(f"The employee_id of this person is {emp_id}")

class depatment(employee):
    def dept(self, department = "IT"):
        print(f"The department of this person is {department}")

class salary(depatment):
    def emp_sal(self, salary = 200000):
        print(f"The salary of this employee is salary {salary}")

a=salary()
a.per_info()
a.emp_info()
a.dept()
a.emp_sal()


```
